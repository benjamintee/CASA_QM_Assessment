```python
import pandas as pd
import geopandas as gpd
import os
```

```python
ONSPD_PATH = "large/ONSPD_FEB_2024_UK.csv"
ONSPD = pd.read_csv(ONSPD_PATH, 
    low_memory=False, 
    encoding="cp1252", 
    usecols=[
        "lsoa21cd",  # lsoa code
        "msoa21cd",  # MSOA code
        "ladcd",     # local authority code
        "ladnm",     # local authority name)
    ])

ONSPD = ONSPD.rename(columns={"lsoa21cd": "LSOA21CD",
                              "msoa21cd": "MSOA21CD", 
                              "ladcd": "LAD24CD",
                              "ladnm": "LAD24NM"})

ONSPD = ONSPD.drop_duplicates(subset=['LSOA21CD']).copy()
ONSPD = ONSPD[ONSPD['LSOA21CD'].str.startswith('E', na=False)].copy()

ONSPD.to_csv('LSOA_MSOA_LAD_Mapping.csv', index=False)
```
